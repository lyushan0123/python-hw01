====== cell1:字符串的乐趣 ========

```python
#====== 1-1 ============
def is_palindrome(s: str) -> bool:
    """
    判断是否回文，忽略空格、忽略大小写；只包含字母数字空格
    """
    # 去掉空格，全部转小写
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]


# 测试
print("is_palindrome测试：")
print(is_palindrome("T A C O cat"))   # True
print(is_palindrome("level"))         # True
print(is_palindrome("hello"))         # False
```

```python
#======= 1-2 =======
def is_abecedarian(s: str) -> bool:
    """
    判断字母是否按字典升序，忽略空格
    """
    cleaned = s.replace(" ", "").lower()
    for i in range(len(cleaned)-1):
        if cleaned[i] > cleaned[i+1]:
            return False
    return True


# 测试
print("is_abecedarian测试：")
print(is_abecedarian("adder"))        # True
print(is_abecedarian("abcd efgh xyz"))# True
print(is_abecedarian("dog"))          # False
```

```python
#======= 1-3 =======
def remove_vowels(s: str) -> str:
    """移除元音 a e i o u，不区分大小写"""
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)


# 测试
print("remove_vowels测试：")
print(remove_vowels("cat"))           # ct
print(remove_vowels("audio"))         # d
```

======= cell2:列表的乐趣 ======

```python
#======= 2-1 =======
def list_reverse(lst):
    """输入列表，返回反转后的列表；非列表抛TypeError"""
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表类型")
    return lst[::-1]

#测试
print("list_reverse: ", list_reverse([1,2,3]))  # [3,2,1]
```

```python
#======= 2-2 =======
def is_sorted(seq) -> bool:
    """
    判断序列是否升序，只遍历一次
    seq支持索引，元素可比较
    """
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

# --------测试--------
print("is_sorted [1,2,3,4] :", is_sorted([1,2,3,4]))
print("is_sorted [1,3,2,4] :", is_sorted([1,3,2,4]))
```

======= cell3:作为向量的元组 ======

```python
#======= 3-1 =======
def vec_scalar_mult(t: tuple, s):
    """
    向量标量乘法 t * s
    t: int/float元组；s int/float
    类型不对抛TypeError
    """
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须为元组（向量）")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是int或者float")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或浮点数")
    return tuple(x * s for x in t)

#测试
v1 = (1,2,3)
print("vec_scalar_mult(v1,2):", vec_scalar_mult(v1, 2))
```

```python
#======= 3-2 =======
def vec_inner_product(t1: tuple, t2: tuple) -> float:
    """向量内积，返回浮点数；非元组抛TypeError，长度不等抛ValueError"""
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组类型向量")
    if len(t1) != len(t2):
        raise ValueError("两个向量维度（长度）必须相等")
    for e in t1 + t2:
        if not isinstance(e, (int, float)):
            raise TypeError("向量元素必须为int/float")
    total = 0.0
    for a,b in zip(t1,t2):
        total += a*b
    return total

#测试
v1 = (1,2,3)
v2 = (4,5,6)
print("vec_inner_product(v1,v2):", vec_inner_product(v1, v2))
```

```python
#======= 3-3 =======
def check_valid_mx(mx) -> bool:
    """
    判断是否合法矩阵：元组的元组
    1.外层每个元素是元组；2.所有内层元组长度一致；3.全部元素int/float
    """
    if not isinstance(mx, tuple):
        return False
    # 每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 所有行长度相等
    row_len = len(mx[0]) if len(mx)>0 else 0
    for row in mx:
        if len(row) != row_len:
            return False
    # 全部元素int/float
    for row in mx:
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

#测试
mat = ((1,2),(3,4))
print("check_valid_mx(mat):", check_valid_mx(mat))
```

```python
#======= 3-4 =======
def mx_vec_mult(mat, vec: tuple) -> tuple:
    """矩阵(元组的元组)乘向量(元组)，返回结果元组
    矩阵每一行和向量做内积
    """
    if not check_valid_mx(mat):
        raise TypeError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(mat[0]) if len(mat)>0 else 0
    if len(vec) != n_col:
        raise ValueError(f"矩阵列数{n_col}与向量长度{len(vec)}不匹配")
    res = []
    for row in mat:
        res.append(vec_inner_product(row, vec))
    return tuple(res)

#测试
mat = ((1,2),(3,4))
print("mx_vec_mult(mat, (1,1)): ", mx_vec_mult(mat, (1,1)))
```

======= cell4:更多向量的乐趣 ======

```python
#======= 4-1 =======
class Vector:
    def __init__(self, dim, elements=None):
        """
        :param dim: 维度（非负整数）
        :param elements: list/tuple 元素序列；不传则默认全部0
        """
        # 校验维度
        if not isinstance(dim, int):
            raise TypeError("维度dim必须为整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        self.dim = dim

        # 不给elements，默认填充0
        if elements is None:
            self.elements = tuple([0.0 for _ in range(dim)])
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements必须是list或者tuple")
            # 全部转为float存为元组
            elem_tuple = tuple(float(x) for x in elements)
            if len(elem_tuple) != dim:
                raise ValueError(f"维度{dim}与元素数量{len(elem_tuple)}不匹配")
            self.elements = elem_tuple

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
```

```python
#======= 4-2 =======
class Vector(Vector):
    def dot(self, other):
        """计算两个Vector的内积，返回浮点数"""
        if not isinstance(other, Vector):
            raise TypeError("dot的参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("两个向量维度必须相等才能求内积")
        s = 0.0
        for a,b in zip(self.elements, other.elements):
            s += a*b
        return s
```

维度和元素应该使用**实例属性**，而不是类属性。

理由：
类属性属于类本身，被该类所有实例对象共享。实例属性属于每一个单独创建出来的对象。
每个Vector向量对象都可以拥有独立的维度与元素。如果把维度和元素设置成类属性，所有向量实例共用同一套数据，修改任意一个向量就会影响全部向量，不符合向量的定义。
所以维度dim和elements必须作为实例属性，在`__init__`中赋值。

```python
#======= 4-3 =======
class Vector(Vector):
    def __eq__(self, other):
        """ == 运算符重载 """
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能做相等比较")
        return self.elements == other.elements

    def __lt__(self, other):
        """ < 字典序比较 """
        if not isinstance(other, Vector):
            raise TypeError("只能Vector与Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较大小")
        return self.elements < other.elements

    def __le__(self, other):
        """ <= """
        return self < other or self == other

    def __gt__(self, other):
        """ > """
        return not self <= other

    def __ge__(self, other):
        """ >= """
        return not self < other
```

```python
#======= 4-4 =======
class Vector(Vector):
    def __mul__(self, other):
        """
        v * x
        - other是数字：标量右乘
        - other是Vector：逐元素相乘
        """
        if isinstance(other, (int, float)):
            new_elems = [x*other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量逐元素相乘维度必须相同")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法仅支持数字或Vector实例")

    def __rmul__(self, scalar):
        """ 标量左乘 2 * v 调用 __rmul__ """
        return self * scalar
```

```python
#======= 4-5 =======
# --------测试Vector类--------
v1 = Vector(2, [2, 3])
v2 = Vector(2, [4, 5])
print("v1:", v1)
print("v2:", v2)
print("v1.dot(v2):", v1.dot(v2))

print("v1 * v2 逐元素相乘：", v1 * v2)
print("v1 * 10 标量右乘：", v1 * 10)
print("3 * v1 标量左乘：", 3 * v1)

print("v1 == Vector(2,[2,3]) :", v1 == Vector(2,[2,3]))
print("Vector(2,(2,4)) < Vector(2,(2,5)) :", Vector(2,(2,4)) < Vector(2,(2,5)))

# 只传维度，元素默认0
v_zero = Vector(3)
print("默认零向量 Vector(3):", v_zero)
```
