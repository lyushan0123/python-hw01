```python
# ========== 实验2‑1 score_level 成绩等级函数 ==========
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

# 测试
print(score_level(95))
print(score_level(75))
```

```python
# ========== 实验2‑3 迭代、递归阶乘 0! =1 ==========
# 迭代版
def fact_iter(n):
    if n < 0:
        raise ValueError("输入非负整数")
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归版
def fact_recur(n):
    if n <0:
        raise ValueError("输入非负整数")
    if n ==0:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_recur(n)}")
```

```python
# ========== 实验2‑4 summary函数，返回最小、最大、平均，元组解包 ==========
def summary(values):
    if len(values) ==0:
        raise ValueError("列表不能为空")
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

lst = [10,20,30,40]
mi, ma, av = summary(lst) #元组解包
print(f"最小{mi},最大{ma},平均{av}")
```

```python
import os
import pickle

#4‑1查看工作目录，创建文件夹
print("当前工作目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)

#准备文本内容
txt_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

stu_path = os.path.join(data_dir, "students.txt")
#4‑2写入文件
with open(stu_path, "w", encoding="utf‑8") as f:
    f.write(txt_content)

#读取文件
with open(stu_path, "r", encoding="utf‑8") as f:
    lines = f.readlines()
print("原始读取行数", len(lines))
#跳过表头
content_lines = lines[1:]

#4‑3解析函数
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) !=2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    name = name.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str.strip())
    except ValueError:
        raise ValueError("成绩不是整数")
    if not (0 <= score <= 100):
        raise ValueError("成绩超出0‑100")
    return {"name":name, "score":score}

#4‑4逐行处理，捕获异常
valid_records = []
error_info = []
for lineno, line in enumerate(content_lines, start=2):
    try:
        rec = parse_student(line)
        valid_records.append(rec)
    except ValueError as e:
        error_info.append({"line":lineno, "err":str(e)})

print("有效记录", valid_records)
print("错误记录", error_info)

#4‑5写report.txt，演示 w / a
valid_num = len(valid_records)
invalid_num = len(error_info)
if valid_num>0:
    avg_s = sum(x["score"] for x in valid_records)/valid_num
    pass_c = sum(1 for x in valid_records if x["score"]>=60)
    pass_r = pass_c / valid_num
else:
    avg_s = 0
    pass_r =0

report_path = os.path.join(data_dir, "report.txt")
# w模式：覆盖原有全部内容
with open(report_path, "w", encoding="utf‑8") as f:
    f.write(f"有效记录:{valid_num}\n无效记录:{invalid_num}\n")
    f.write(f"平均分:{avg_s:.2f}\n及格率:{pass_r:.1%}\n")

# a模式：追加，在文件末尾新增文字
with open(report_path, "a", encoding="utf‑8") as f:
    f.write("这是a模式追加的一行\n")

#4‑6 pickle保存读取
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)
print("pickle读取是否相等：", loaded_data == valid_records)

#4‑8捕获不存在文件异常
try:
    with open("no_exist.txt","r") as f:
        pass
except FileNotFoundError as e:
    print("捕获文件不存在异常：", e)
```

```python
import copy

#5‑1 Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name},score={self.score})"

#5‑2 由实验4有效记录生成对象
obj_list = []
for r in valid_records:
    s = Student(r["name"], r["score"])
    obj_list.append(s)
    print(s, "及格" if s.is_passed() else "不及格")

#5‑3子类GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g = GraduateStudent("LiHua",88,"AI")
print(g)

#5‑4直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy
original = {"name": "Alice", "scores": [85, 90]}

#测试1 直接赋值
o1 = original
o1["scores"][0] = 999
print("直接赋值后原对象", original)
original = {"name": "Alice", "scores": [85, 90]} #重置原始

#测试2 浅拷贝
o2 = copy.copy(original)
o2["scores"][0] = 999
print("浅拷贝后原对象", original)
original = {"name": "Alice", "scores": [85, 90]} #重置原始

#测试3深拷贝
o3 = copy.deepcopy(original)
o3["scores"][0] = 999
print("深拷贝后原对象", original)
```
