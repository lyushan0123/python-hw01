```python
#========欧几里得算法========

def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数据
print("gcd(2018,2019) =", gcd(2018,2019))
print("gcd(1200,300) =", gcd(1200,300))
print("gcd(5040,60) =", gcd(5040,60))

# 测试负数
print("\n负数测试 gcd(-12, -8):", gcd(-12,-8))
print("负数测试 gcd(-12, 8):", gcd(-12,8))
```

```python
#=========近似欧拉数e========

import math

# 1.极限形式 euler_limit
def euler_limit(n):
    return (1.0 + 1.0/n)**n

# 2.级数前n项和 euler_infinite_sum
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
    return total

#3. euler_approx，误差小于epsilon
def euler_approx(epsilon):
    real_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - real_e) >= epsilon:
        if k == 0:
            fact =1
        else:
            fact *=k
        total += 1.0/fact
        k +=1
    return total

#4.print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")


# 测试
print("euler_limit(10000):", euler_limit(10000))
print("euler_infinite_sum(0):", euler_infinite_sum(0))
print("euler_infinite_sum(1):", euler_infinite_sum(1))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("真实e math.exp(1):", math.exp(1))
print("euler_approx(1e-4):", euler_approx(1e-4))
print("\n==== print_euler_sum_table(6) ====")
print_euler_sum_table(6)
```

```python
#=======测试整数属性======

# 1.is_power_of_2 判断是否为2的幂，只用除法、取模
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

#2.通用 is_power(b,n) n是否是b的幂
def is_power(b, n):
    while n > 1:
        if n % b !=0:
            return False
        n = n // b
    return True


#测试
print("is_power_of_2(1):", is_power_of_2(1))
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(9):", is_power_of_2(9))

print("\nis_power(3,9):", is_power(3,9))
print("is_power(2,64):", is_power(2,64))
print("is_power(5,13):", is_power(5,13))
```
